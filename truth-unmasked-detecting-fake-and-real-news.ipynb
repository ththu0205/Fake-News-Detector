# <p style="background-color: #6b5b95; font-family: 'Verdana', sans-serif; font-size: 28px; text-align: center; color: #f7f7f7; padding: 20px; border-radius: 50px; text-shadow: 3px 3px 8px rgba(0, 0, 0, 0.6); border: 4px solid #ffcc00; box-shadow: 0 4px 15px rgba(0, 0, 0, 0.5);">Lowercasing</p>


* `Lowercasing` is the process of converting all uppercase letters of a text to the lowercase. 
* The function for lower case defined here converts all the uppercase letters of the text to the lowercase letters
* `As the words written in uppercase & lowercase or a combination of both are treated as different entities in most of text-processing algorithms so it is necessary to ensure that they are treated as the same entity inorder to get the good performence of the Model`
* Hence, It is very essential step of working with textual Data and especially for NLP tasks to Standerdize and Clean the Data and maintains the consistency in the Text.
* It helps the model to focus on meaningful patterns of the Data rather than being misled by different capitalizations of the Text.
* Converting to Lowercase also helps in tokenization which is the process of breaking down the text into individual words or tokens.
* The function `lowercase_text` is defined which takes one argument of the text.
* The function applies the Python’s built-in `.lower()` method to the input text and convert the text into the lowercase.
* `.lower()` returns a lowercase string or text and it ensures all letters in the text are uniformly converted.

```python
# Convert all text to lowercase
def lowercase_text(text):
    return text.lower()

df['Text'] = df['Text'].apply(lowercase_text)

```

# <p style="background-color: #6b5b95; font-family: 'Verdana', sans-serif; font-size: 28px; text-align: center; color: #f7f7f7; padding: 20px; border-radius: 50px; text-shadow: 3px 3px 8px rgba(0, 0, 0, 0.6); border: 4px solid #ffcc00; box-shadow: 0 4px 15px rgba(0, 0, 0, 0.5);">Remove HTML Tags</p>


* `Removing the HTML tags from a string means that stripping out the markup elements from the string, and leave only the plain text content. This is typically done to standerdize the text and ensures that no unwanted tags remains in the string.`
* The function defined here called the remove_html_tags is used to remove any HTML tags present in the text.
* This function basically uses the BeautifulSoup library from the bs4 module and parse the input text.
* The `"html.parser"` tells BeautifulSoup to use its built-in HTML parser which is efficient for extracting text from HTML content.
* `.get_text()` extracts and returns the plain text from the HTML content and strips away the HTML tags.
* `remove_html_tags` function applies to every entry in the 'Text' column of the DataFrame.
* And Using the `.apply()` method ensures that each string in the 'Text' column is cleaned from HTML tags, and the result is stored back in the same column.
* When working with textual data that may originate from web pages the user comments and other sources with embedded HTML to remove HTML tags is necessary to clean the Data and use for modeling.
* `Working with Machine Learning or NLP models removing HTML tags helps to reduce the noise in the data and hence it allows the Model to focus only on the relevant and actual content and not the formatting`


```python
# Remove HTML tags
def remove_html_tags(text):
    return BeautifulSoup(text, "html.parser").get_text()

df['Text'] = df['Text'].apply(remove_html_tags)

```

# <p style="background-color: #6b5b95; font-family: 'Verdana', sans-serif; font-size: 28px; text-align: center; color: #f7f7f7; padding: 20px; border-radius: 50px; text-shadow: 3px 3px 8px rgba(0, 0, 0, 0.6); border: 4px solid #ffcc00; box-shadow: 0 4px 15px rgba(0, 0, 0, 0.5);">Remove URLs</p>


* This function is used to remove URLs from a Text of the Data.
* `re.sub(`) function from the re (regular expressions) module is used to search for URL patterns in the text and replace them with an empty string ('') and hence, effectively remove them.
* The pattern used (r'http\S+|www\S+|https\S+') here is a regular expression that matches the http or https links (http\S+ and https\S+ match any URL starting with http or https).or www links (www\S+ matches URLs starting with www).
* The `\S+` matches any sequence of non-whitespace characters of the URL.
* `flags=re.MULTILINE` ensure that URLs appearing across multiple lines are removed from the Text.
* As, URLs do not add any semantic meaning to text data Hence, Removing  the URLs simplifies the text, and make the modeling part efficient
* It reduces the nise present in the Data
* `remove_urls` function apply to each entry in the 'Text' column of the DataFrame.
* `Using .apply()` method allows the function to process each text entry, detect and remove any URLs, and update the 'Text' column with the cleaned text.


```python
# Remove URLs
def remove_urls(text):
    return re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)

df['Text'] = df['Text'].apply(remove_urls)
```

# <p style="background-color: #6b5b95; font-family: 'Verdana', sans-serif; font-size: 28px; text-align: center; color: #f7f7f7; padding: 20px; border-radius: 50px; text-shadow: 3px 3px 8px rgba(0, 0, 0, 0.6); border: 4px solid #ffcc00; box-shadow: 0 4px 15px rgba(0, 0, 0, 0.5);">Handling Chat Words</p>


* Handling the chat words helps to clean the textual data by eliminating common abbreviations and slang that do not contribute much value to text analysis and also causes noise in text Data.
* It simplifies the Text and helps to improve the text analysis, reduces noise and Hence, make the Dataset more uniform.
* This step is particularly useful when dealing with informal text data such as social media, chats etc. and also when we need to improve the quality of the text for further processing and modeling.
* The function `remove_chat_words` serves to enhance the readability and clarity of textual data and replace the common chat abbreviations with their expanded forms. This transformation make the text more clean, accessible, and suitable to use especially in NLP and data analysis tasks.
* The loop defined here iterate over each abbreviation in the chat_words dictionary.
* The function is applied to each entry in the Text column of the DataFrame df.
* `.apply()` method allows us to process each text entry in the Text column and hence replace any chat words with their expanded forms.
* It Helps to ensure the consistency accross the Text

```python
def remove_chat_words(text):
    chat_words = r'\b(?:lol|brb|omg|btw|idk|tbh|smh|fyi|imo|lmao|bff|ttyl)\b'
    return re.sub(chat_words, '', text, flags=re.IGNORECASE)

df['Text'] = df['Text'].apply(remove_chat_words)
```

# <p style="background-color: #6b5b95; font-family: 'Verdana', sans-serif; font-size: 28px; text-align: center; color: #f7f7f7; padding: 20px; border-radius: 50px; text-shadow: 3px 3px 8px rgba(0, 0, 0, 0.6); border: 4px solid #ffcc00; box-shadow: 0 4px 15px rgba(0, 0, 0, 0.5);">Handling Stop Words</p>


* In  the NLP tasks the stop words are inconsequential words with little value in helping processors answer queries. In order to clean and standerdize the textual Data it typically needs to manually filter out words that would help to select relevant content of the Data. 
* The code firstly imports the nltk (Natural Language Toolkit) library as a powerful toolkit helps to work with human language data.
* The `nltk.download('stopwords')` command used here downloads list of common stopwords for various languages such as English. 
* `Stopwords` are the words that are often filtered out for text processing tasks to make Data clean and Standerdize for modeling.
* The function remove_stopwords takes a string text as input and it splits the text into individual words using `text.split()`.
* A list comprehension is used to iterate through each word of the text by using the code.  If the word is not in the stop_words,then it is included in the new list.
* `join()` is then used to combine the remaining words back into a single string and effectively removing the stopwords.
* Hence, `remove_stopwords` function apply to each entry in the Text column of the DataFrame.
* Using `.apply()`  method helps to ensure that each text entry is processed and the updated the text without stopwords and replaces the original text in the DataFrame.
* It Helps to improve the Dta Quality, Reduce Noise, Enhance Computational Efficiency

```python
# Download stopwords from nltk
import nltk
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    return ' '.join([word for word in text.split() if word not in stop_words])

df['Text'] = df['Text'].apply(remove_stopwords)
```

# <p style="background-color: #6b5b95; font-family: 'Verdana', sans-serif; font-size: 28px; text-align: center; color: #f7f7f7; padding: 20px; border-radius: 50px; text-shadow: 3px 3px 8px rgba(0, 0, 0, 0.6); border: 4px solid #ffcc00; box-shadow: 0 4px 15px rgba(0, 0, 0, 0.5);">Removing Punctuation</p>


* The code provides effectively removes the punctuation from the text in the DataFrame's Text column. 
* By eliminating the characters such as commas, periods, exclamation marks etc the text becomes cleaner and more standardized.
* while this preprocessing step is crucial for enhancing the quality and relevance of the textual data, it ensures the subsequent analyses focus on the meaningful content and contribute to more accurate insights can be taken from the data.
* The function remove_punctuation is defined to take a string text as input.
* Hencce, inside the function, `str.maketrans('', '', string.punctuation)` helps to create a translation table that maps each punctuation character to None, and indicates that these characters should be removed from the text.

# <p style="background-color: #6b5b95; font-family: 'Verdana', sans-serif; font-size: 28px; text-align: center; color: #f7f7f7; padding: 20px; border-radius: 50px; text-shadow: 3px 3px 8px rgba(0, 0, 0, 0.6); border: 4px solid #ffcc00; box-shadow: 0 4px 15px rgba(0, 0, 0, 0.5);">Removing Extra White Space</p>


* The code helps to effectively removes extra whitespace from the text in the DataFrame's Text column. * By replacing multiple spaces of the Text Data with just one space and removing any spaces at the beginning or end of the text we can make the text clearer and easier to read
* This step is very important for improving the overall quality and uniformity of the textual data.
* This function `remove_extra_whitespace` is used to take a string text as input.
* Inside this function `re.sub(r'\s+', ' ', text)` uses a regular expression which helps to find occurrences of one or more whitespace characters such as spaces, tabs, newlines, etc. and replaces them with a single space (' ') in the textual Data.
* `\s+` matches sequence of whitespace characters while `re.sub()` replaces it with a single space.
* The `strip()` method called on the result to remove any leading or trailing whitespace present in textual Data Hence, it ensures that the text does not start or end with unnecessary spaces.

```python
def remove_extra_whitespace(text):
    return re.sub(r'\s+', ' ', text).strip()

df['Text'] = df['Text'].apply(remove_extra_whitespace)
```

# <p style="background-color: #6b5b95; font-family: 'Verdana', sans-serif; font-size: 28px; text-align: center; color: #f7f7f7; padding: 20px; border-radius: 50px; text-shadow: 3px 3px 8px rgba(0, 0, 0, 0.6); border: 4px solid #ffcc00; box-shadow: 0 4px 15px rgba(0, 0, 0, 0.5);">Handling Emojis</p>


* This code is used to effectively removes the emojis from the text in the DataFrame's Text column.
* By eliminating these characters such as emojis, the text becomes cleaner and more standardized.
* This preprocessing step is important for enhancing the quality of the textual data and it ensures that the subsequent analyses can focus on the meaningful content only and result in more accurate insights and modeling from the data.
* The function `remove_emojis` is defined here to take a string text as the input.
* Inside this function the emoji_pattern is created using `re.compile()` to define a regular expression that matches a wide range of emoji characters present in the text.
* The character ranges cover various categories such as emotions, symbolsetc
* The emoji_pattern `.sub(r'', text)` method replaces any matched emojis in the input text with an empty string.
* remove_emojis function then applies to each entry in the Text column of the DataFrame.
* Using `.apply()` method helps the text to be processed and the updated text with emojis removed and replace the original text in the DataFrame.
* It Standardize the Text Data, Enhance the Text Analysis, Reduce Noise and Improve Readability

```python
def remove_emojis(text):
    emoji_pattern = re.compile("["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map symbols
        u"\U0001F700-\U0001F77F"  # alchemical symbols
        u"\U0001F780-\U0001F7FF"  # Geometric Shapes Extended
        u"\U0001F800-\U0001F8FF"  # Supplemental Arrows-C
        u"\U0001F900-\U0001F9FF"  # Supplemental Symbols and Pictographs
        u"\U0001FA00-\U0001FAFF"  # Chess Symbols
        u"\U00002702-\U000027B0"  # Dingbats
        "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)

df['Text'] = df['Text'].apply(remove_emojis)
```

# <p style="background-color: #6b5b95; font-family: 'Verdana', sans-serif; font-size: 28px; text-align: center; color: #f7f7f7; padding: 20px; border-radius: 50px; text-shadow: 3px 3px 8px rgba(0, 0, 0, 0.6); border: 4px solid #ffcc00; box-shadow: 0 4px 15px rgba(0, 0, 0, 0.5);">Stemming</p>


* `Stemming` is the process of stripping the affixes from words or text to obtain stemmed word strings
* The idea behind the  stemming is to take away the different endings of words and finds the most basic part of the text which is the “stem.”
### `Various Purpose of Stemming:`

`1. Reducing Word Variability:`
Stemming helps to reduce the number of unique words in the textual data by applying different forms of a word into a single representation. This helps to make the dataset less sparse and more manageable for analysis and modeling.

`2. Improving Text Analysis:`
Standardizing the words to their root forms helps to ensure that the words with similar meanings are treated as the same word. 

`3. Enhancing Search and Retrieval:`
Stemming is mostly used in search engines to match queries with documents. 

`4. Reducing Dimensionality:`
The fewer unique terms (after stemming) can helps to reduce the dimensionality which leads to a faster processing and better model performance.

* `nltk.download('punkt')` command is used to downloads the code and necessary resources from the Natural Language Toolkit (NLTK) for tokenization process which is the process of splitting text into individual words or tokens.
* `stemmer = PorterStemmer()` is usually intiated to create an object named stemmer that will be used for stemming the words in the text.
* stemming function ususally takes a string text as the input.
* and the input text is then split into individual words using the `split() method` creates a list of words.
* Then A list comprehension iterates over each word, applying the stemmer.stem() method to reduce each word to its stem.
* The stemmed words are then joined back together into a single string using ' '.join(), and this cleaned text is then returned.
* .apply() function applies the stemming to the Text column of the DataFrame.

```python
# Download necessary resources for stemming
nltk.download('punkt')
stemmer = PorterStemmer()

def stemming(text):
    words = text.split()
    return ' '.join([stemmer.stem(word) for word in words])

df['Text'] = df['Text'].apply(stemming)
# Show the preprocessed text
df.head()
```

This visualization helps to illustrate the differences in vocabulary and themes between the two categories and provides the valuable insights about the nature of the content and helps in text analysis.
* Word clouds are very useful for visualizing the textual data and making themvery important for exploratory data analysis.
### `Purpose of Creating Word Clouds:`

   ` 1. Visualizing the Word Frequency:`
Word clouds is very useful and effective visual representation of the common words in a text corpus.

  ` 2. Identifying Trends:`
By comparing word clouds of different categories such as real vs. fake news we can easily identify differences in vocabulary, tone, and style which can indicate broader trends and biases.

  ` 3. Enhancing Text Analysis:`
Word clouds are very useful in exploratory data analysis hence, it allows researchers to quickly take insights about the content of a text dataset before conducting more detailed analyses.
#### `Word Cloud Representation:`

   `1. Real News Word Cloud: Words like "Said", "Donald" , "trump", "hous","presid", "Unit", "State" may appear larger, indicating their frequency in the dataset.`
   
  `2. Fake News Word Cloud: Words like "republican", "trump", "Donald", "even", "say", "peopl", "call" might be more prominent, suggesting a different focus or style.`


```python
# Create a function to generate word clouds
def plot_word_cloud(data, title):
    wordcloud = WordCloud(width=800, height=400, background_color='white').generate(' '.join(data))
    
    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis('off')
    plt.title(title)
    plt.show()

# Generate word clouds for Real and Fake news
plot_word_cloud(df[df['label'] == 'Real']['Text'], 'Word Cloud for Real News')
plot_word_cloud(df[df['label'] == 'Fake']['Text'], 'Word Cloud for Fake News')
```

```python
# Initialize the Sentiment Intensity Analyzer
nltk.download('vader_lexicon')
sia = SentimentIntensityAnalyzer()
```

```python
# Function to get sentiment scores
def get_sentiment(text):
    return sia.polarity_scores(text)['compound']

# Apply sentiment analysis to each article
df['sentiment'] = df['Text'].apply(get_sentiment)
```

# <p style="background-color: #6b5b95; font-family: 'Verdana', sans-serif; font-size: 28px; text-align: center; color: #f7f7f7; padding: 20px; border-radius: 50px; text-shadow: 3px 3px 8px rgba(0, 0, 0, 0.6); border: 4px solid #ffcc00; box-shadow: 0 4px 15px rgba(0, 0, 0, 0.5);">Visualize the Sentiment Polarity Distribution by VADER</p>

```python
# Calculate mean, max, and min sentiment scores for each label
mean_sentiment = df.groupby('label')['sentiment'].mean().reset_index()
max_sentiment = df.groupby('label')['sentiment'].max().reset_index()
min_sentiment = df.groupby('label')['sentiment'].min().reset_index()

# Merge mean, max, and min into a single DataFrame for easier annotation
summary_sentiment = pd.merge(mean_sentiment, max_sentiment, on='label', suffixes=('_mean', '_max'))
summary_sentiment = pd.merge(summary_sentiment, min_sentiment, on='label')
summary_sentiment.rename(columns={'sentiment': 'sentiment_min'}, inplace=True)

# Print mean, max, and min scores
for index, row in summary_sentiment.iterrows():
    print(f"Label: {row['label']}, Mean Sentiment: {row['sentiment_mean']:.2f}, Max Sentiment: {row['sentiment_max']:.2f}, Min Sentiment: {row['sentiment_min']:.2f}")

# Create a violin plot for sentiment scores with different colors and legends
fig = px.violin(df, 
                 x='label', 
                 y='sentiment', 
                 box=True, 
                 points='all', 
                 title='Sentiment Polarity Distribution by Label',
                 color='label',  # Color by label
                 color_discrete_sequence=['#636EFA', '#EF553B'])  # Custom colors

# Update layout to add legends and improve aesthetics
fig.update_traces(meanline_visible=True)  # Show mean line
fig.update_layout(yaxis_title='Sentiment Score',
                  xaxis_title='News Category',
                  legend_title='Label',
                  legend=dict(x=0.85, y=0.9))  # Position the legend

fig.show()
```

```python
def get_top_n_words(corpus, n=10):
    words = ' '.join(corpus).split()
    common_words = Counter(words)
    return common_words.most_common(n)
```

```python
def plot_top_n_words(data, title, n=10):
    top_words = get_top_n_words(data['Text'], n)
    words, counts = zip(*top_words)
    
    # Create a bar plot with different colors and annotations
    fig = px.bar(x=words, y=counts, title=title, 
                 labels={'x': 'Words', 'y': 'Counts'},
                 color=counts,  # Use counts to vary color
                 color_continuous_scale=px.colors.sequential.Viridis)  # Choose a color scale

    # Add text annotations to the bars
    fig.update_traces(text=counts, textposition='outside')
    
    # Update layout to improve aesthetics
    fig.update_layout(
        xaxis_title='Words',
        yaxis_title='Counts',
        showlegend=False,
        plot_bgcolor='rgba(0,0,0,0)',
        title_font=dict(size=18),
        xaxis=dict(tickangle=45)  # Rotate x-axis labels for better readability
    )
    
    fig.show()

# Plot top 10 words for Real and Fake news
plot_top_n_words(df[df['label'] == 'Real'], 'Top 10 Words in Real News', n=10)
plot_top_n_words(df[df['label'] == 'Fake'], 'Top 10 Words in Fake News', n=10)
```

```python
# Splitting the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(df['Text'], df['label'], test_size=0.2, random_state=42)

# Create a Count Vectorizer for Bag of Words
vectorizer = CountVectorizer()
X_train_bow = vectorizer.fit_transform(X_train)
X_test_bow = vectorizer.transform(X_test)
```

# <p style="background-color: #6b5b95; font-family: 'Verdana', sans-serif; font-size: 28px; text-align: center; color: #f7f7f7; padding: 20px; border-radius: 50px; text-shadow: 3px 3px 8px rgba(0, 0, 0, 0.6); border: 4px solid #ffcc00; box-shadow: 0 4px 15px rgba(0, 0, 0, 0.5);">Display Top Real and FAKE News Words</p>

```python
# Separate the DataFrame into real and fake news
real_news = df[df['label'] == 'Real']['Text']
fake_news = df[df['label'] == 'Fake']['Text']

# Vectorization for Real News
vectorizer_real = CountVectorizer()
X_real_bow = vectorizer_real.fit_transform(real_news)

# Get feature names (words) for real news
feature_names_real = vectorizer_real.get_feature_names_out()

# Get the sums of each word in real news articles
word_counts_real = X_real_bow.toarray().sum(axis=0)

# Create a DataFrame for words and their counts in real news
words_df_real = pd.DataFrame({'word': feature_names_real, 'count': word_counts_real})

# Vectorization for Fake News
vectorizer_fake = CountVectorizer()
X_fake_bow = vectorizer_fake.fit_transform(fake_news)

# Get feature names (words) for fake news
feature_names_fake = vectorizer_fake.get_feature_names_out()

# Get the sums of each word in fake news articles
word_counts_fake = X_fake_bow.toarray().sum(axis=0)

# Create a DataFrame for words and their counts in fake news
words_df_fake = pd.DataFrame({'word': feature_names_fake, 'count': word_counts_fake})

# Get top 10 words for Real and Fake news
top_real_words = words_df_real.sort_values(by='count', ascending=False).head(15)
top_fake_words = words_df_fake.sort_values(by='count', ascending=False).head(15)

# Print the top words
print("Top 15 Words in Real News:")
print(top_real_words)
print("===============================")

print("\nTop 15 Words in Fake News:")
print(top_fake_words)
```

```python
# Initialize models
nb_model = MultinomialNB()
svm_model = SVC(probability=True)

# Train the models
nb_model.fit(X_train_bow, y_train)
svm_model.fit(X_train_bow, y_train)
from sklearn.metrics import accuracy_score
import numpy as np

# Assuming y_test contains the true labels for your test set

# Make predictions
nb_predictions = nb_model.predict(X_test_bow)
svm_predictions = svm_model.predict(X_test_bow)

# Calculate accuracy for Naive Bayes
nb_accuracy = accuracy_score(y_test, nb_predictions)
print(f"Naive Bayes Model Accuracy: {nb_accuracy:.4f}")

# Calculate accuracy for SVM
svm_accuracy = accuracy_score(y_test, svm_predictions)
print(f"SVM Model Accuracy: {svm_accuracy:.4f}")
```

```python
def plot_confusion_matrix(y_true, y_pred, model_name):
    """Plot confusion matrix."""
    # Compute confusion matrix
    conf_matrix = confusion_matrix(y_true, y_pred)
    
    # Create a heatmap for the confusion matrix
    plt.figure(figsize=(6, 4))
    sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', 
                xticklabels=['Fake', 'Real'], 
                yticklabels=['Fake', 'Real'])
    plt.title(f'Confusion Matrix for {model_name}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()

# Now you can plot the confusion matrices for each model in separate cells.

# For Naive Bayes
plot_confusion_matrix(y_test, nb_predictions, "Naive Bayes")

# For SVM
plot_confusion_matrix(y_test, svm_predictions, "SVM")
```

```python
import pandas as pd
import numpy as np
import torch
from torch.utils.data import Dataset
from sklearn.model_selection import train_test_split
from transformers import BertTokenizer, BertForSequenceClassification, Trainer, TrainingArguments, EarlyStoppingCallback
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Ensure the text column is named correctly in the DataFrame
X = df['Text'].astype(str).tolist()
y = df['label'].values  # Ensure that labels are in a suitable format

# Convert labels to numeric format (if not already numeric)
y = pd.factorize(y)[0]

# Data Augmentation (implementing better logic, such as synonyms, paraphrasing)
def augment_text(text):
    # Here, you could use libraries like nlpaug to replace words with synonyms
    return text  # Replace with actual augmentation logic

# Augment the dataset
X = [augment_text(text) for text in X]

# Stratified Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, stratify=y, random_state=42)

# Tokenization
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Custom Dataset Class
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, return_tensors='pt', max_length=128)
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        item = {key: val[idx] for key, val in self.encodings.items()}
        item['labels'] = self.labels[idx]
        return item

    def __len__(self):
        return len(self.labels)

# Convert the training and test data into dataset
train_dataset = TextDataset(X_train, y_train)
test_dataset = TextDataset(X_test, y_test)

# Model definition with adjusted dropout and regularization
from transformers import BertConfig
config = BertConfig.from_pretrained('bert-base-uncased', num_labels=2, hidden_dropout_prob=0.5, attention_probs_dropout_prob=0.5)  # Increased dropout rates
bert_model = BertForSequenceClassification.from_pretrained('bert-base-uncased', config=config)

# Define training arguments with adjusted regularization and batch size
training_args = TrainingArguments(
    output_dir='./results',
    num_train_epochs=3,               # Reduced number of epochs
    per_device_train_batch_size=4,    # Lower batch size for better generalization
    per_device_eval_batch_size=16,    
    warmup_steps=300,
    weight_decay=0.4,                 # Increased weight decay for regularization
    logging_dir='./logs',
    logging_steps=10,
    gradient_accumulation_steps=2,    
    fp16=True,                        
    evaluation_strategy="epoch",      
    save_strategy="epoch",            
    load_best_model_at_end=True,      
    lr_scheduler_type="linear",       
    learning_rate=2e-5,               # Keep adjusted learning rate
)

# Early stopping callback with patience for overfitting prevention
early_stopping = EarlyStoppingCallback(early_stopping_patience=2)  

# Trainer initialization
trainer = Trainer(
    model=bert_model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    callbacks=[early_stopping],
)

# Train the model
trainer.train()

# Making predictions
predictions = trainer.predict(test_dataset)

# Get the predicted labels
predicted_labels = np.argmax(predictions.predictions, axis=1)

# Accuracy
accuracy = accuracy_score(y_test, predicted_labels)
print(f"Accuracy: {accuracy}")

# Classification report
print("Classification Report:")
print(classification_report(y_test, predicted_labels))

# Confusion Matrix
conf_matrix = confusion_matrix(y_test, predicted_labels)

# Plot confusion matrix
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', xticklabels=['Fake', 'Real'], yticklabels=['Fake', 'Real'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```
